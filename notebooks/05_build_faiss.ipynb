```python
import faiss
import numpy as np
from pathlib import Path

# Paths
EMBED_FILE_ST = Path("embeddings/st_embeddings.npy")
EMBED_FILE_OPENAI = Path("embeddings/openai_embeddings.npy")

INDEX_DIR = Path("indices")
INDEX_DIR.mkdir(exist_ok=True)

# Load embeddings
emb_st = np.load(EMBED_FILE_ST).astype("float32")
emb_openai = np.load(EMBED_FILE_OPENAI).astype("float32")

# ----------------------------
# SentenceTransformer FAISS
# ----------------------------
dim_st = emb_st.shape[1]
index_st = faiss.IndexFlatL2(dim_st)
index_st.add(emb_st)

faiss.write_index(index_st, str(INDEX_DIR / "faiss_index_st.index"))

print(f"ST index built | vectors={index_st.ntotal} | dim={dim_st}")

# ----------------------------
# OpenAI FAISS
# ----------------------------
dim_oa = emb_openai.shape[1]
index_oa = faiss.IndexFlatL2(dim_oa)
index_oa.add(emb_openai)

faiss.write_index(index_oa, str(INDEX_DIR / "faiss_index_openai.index"))

print(f"OpenAI index built | vectors={index_oa.ntotal} | dim={dim_oa}")
print("FAISS indices saved successfully!")
```
