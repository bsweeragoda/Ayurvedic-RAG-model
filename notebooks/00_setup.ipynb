```python
# -------------------------------------------------------------
# 00_setup.ipynb – Initial environment configuration
# -------------------------------------------------------------

import os
import sys
import subprocess
import yaml
from pathlib import Path

print("Python version:", sys.version)
print("Working directory:", os.getcwd())


# -------------------------------------------------------------
# 1️ Install Required Packages (Safe Re-Install Check)
# -------------------------------------------------------------
required_packages = [
    "sentence-transformers",
    "faiss-cpu",
    "pandas",
    "numpy",
    "pyyaml",
    "openai",
]

def install_if_missing(pkg):
    try:
        __import__(pkg.split("-")[0])   # crude but works for this context
        print(f"✓ {pkg} already installed")
    except ImportError:
        print(f" Installing {pkg}...")
        subprocess.check_call([sys.executable, "-m", "pip", "install", pkg])

for p in required_packages:
    install_if_missing(p)


# -------------------------------------------------------------
# 2️ Load Configuration File
# -------------------------------------------------------------
config_path = Path("config.yaml")

if not config_path.exists():
    raise FileNotFoundError("config.yaml not found. Please create it in project root.")

with open(config_path, "r", encoding="utf-8") as f:
    config = yaml.safe_load(f)

print("\nLoaded config:")
config


# -------------------------------------------------------------
# 3️ Resolve Important Paths
# -------------------------------------------------------------
ROOT = Path(config["paths"]["root"]).resolve()
DATA_DIR = Path(config["paths"]["data"]).resolve()
RAW_DIR = DATA_DIR / "raw"
CHUNKS_FILE = DATA_DIR / "chunks.jsonl"
MASTER_FILE = DATA_DIR / "master.jsonl"

EMB_DIR = Path(config["paths"]["embeddings"]).resolve()
IDX_DIR = Path(config["paths"]["indices"]).resolve()

print("\nResolved Paths:")
print("Root:", ROOT)
print("Data:", DATA_DIR)
print("Raw Data:", RAW_DIR)
print("Embeddings:", EMB_DIR)
print("Indices:", IDX_DIR)


# -------------------------------------------------------------
# 4️ Create Required Folders If Missing
# -------------------------------------------------------------
for d in [DATA_DIR, RAW_DIR, EMB_DIR, IDX_DIR]:
    d.mkdir(parents=True, exist_ok=True)
    print(f"✓ Created or exists → {d}")


# -------------------------------------------------------------
# 5️ Test Imports (Guarantees Everything Works)
# -------------------------------------------------------------
print("\nTesting module imports...")

try:
    import numpy as np
    import pandas as pd
    from sentence_transformers import SentenceTransformer
    import faiss
    import openai

    print("✓ numpy, pandas, ST model, FAISS, openai imported successfully")
except Exception as e:
    print(" Import error:", e)
    raise e


# -------------------------------------------------------------
# 6️ Final Status Message
# -------------------------------------------------------------
print("\nEnvironment setup complete!")
print("You can now run → 01_preprocess.ipynb")
```
