```python
# 03_embed_sentence_transformers.ipynb
# -------------------------------------------------------
# Embed chunks using SentenceTransformers
# -------------------------------------------------------

import json
import numpy as np
from pathlib import Path
from sentence_transformers import SentenceTransformer

CHUNKS_FILE = Path("data/chunks.jsonl")
EMBED_FILE = Path("embeddings/st_embeddings.npy")
META_FILE = Path("embeddings/metadata.json")

model = SentenceTransformer("all-MiniLM-L6-v2")

texts, metadata = [], []

with CHUNKS_FILE.open(encoding="utf-8") as f:
    for line in f:
        rec = json.loads(line)
        texts.append(rec["text"])
        metadata.append({
            "chunk_id": rec["doc_id"],
            "lang": rec["lang"],
            "disease_name_en": rec.get("disease_name_en"),
            "disease_name_si": rec.get("disease_name_si")
        })

embeddings = model.encode(texts, show_progress_bar=True, batch_size=32)
EMBED_FILE.parent.mkdir(exist_ok=True)
np.save(EMBED_FILE, embeddings)

with META_FILE.open("w", encoding="utf-8") as f:
    json.dump(metadata, f, ensure_ascii=False, indent=2)

print(f"Saved SentenceTransformer embeddings: {EMBED_FILE}")
print(f"Saved metadata: {META_FILE}")
```
