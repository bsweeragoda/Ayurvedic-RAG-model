```python
# 03_neo4j_build_faiss.ipynb
# -------------------------------------------
# Build FAISS indices for Neo4j embeddings
# -------------------------------------------

import faiss
import numpy as np
from pathlib import Path

BASE_DIR = Path("graph")
EMB_DIR = BASE_DIR / "embeddings"
IDX_DIR = BASE_DIR / "indices"
IDX_DIR.mkdir(exist_ok=True)

# -------- SentenceTransformer --------
st_vecs = np.load(EMB_DIR / "neo4j_st.npy")
dim = st_vecs.shape[1]

index_st = faiss.IndexFlatL2(dim)
index_st.add(st_vecs)

faiss.write_index(index_st, str(IDX_DIR / "faiss_index_st.index"))

# -------- OpenAI --------
oa_vecs = np.load(EMB_DIR / "neo4j_openai.npy")

index_oa = faiss.IndexFlatL2(oa_vecs.shape[1])
index_oa.add(oa_vecs)

faiss.write_index(index_oa, str(IDX_DIR / "faiss_index_openai.index"))

print("FAISS indices built")
```
