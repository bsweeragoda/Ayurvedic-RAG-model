```python
# 02_neo4j_embed_openai.ipynb
# -------------------------------------------
# Neo4j → Text → OpenAI embeddings
# -------------------------------------------

import json
import numpy as np
import pandas as pd
from openai import OpenAI
from pathlib import Path
from dotenv import load_dotenv
import os
from tqdm import tqdm

# -------- Config --------
load_dotenv()
client = OpenAI(api_key=os.getenv("OPENAI_API_KEY"))

MODEL = "text-embedding-3-small"

BASE_DIR = Path("graph")
DATA_DIR = BASE_DIR / "data"
EMB_DIR = BASE_DIR / "embeddings"

facts = []
with open(DATA_DIR / "neo4j_facts.jsonl", "r", encoding="utf-8") as f:
    for line in f:
        facts.append(json.loads(line)["text"])

# -------- Embed --------
vectors = []

for text in tqdm(facts):
    emb = client.embeddings.create(
        model=MODEL,
        input=text
    ).data[0].embedding
    vectors.append(emb)

vectors = np.array(vectors, dtype="float32")

np.save(EMB_DIR / "neo4j_openai.npy", vectors)

print("OpenAI embeddings saved")
```
